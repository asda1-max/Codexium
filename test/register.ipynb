```python
from cryptography.hazmat.primitives.ciphers.aead import ChaCha20Poly1305
from argon2 import PasswordHasher
from argon2 import hash_password_raw
import random,string
import base64
import uuid
import os
import sqlite3
import math

```

```python
k=40
allowedWords = string.ascii_lowercase + string.digits + string.ascii_uppercase
recovery_code_raw = ''.join(random.choices(allowedWords, k=k))
recovery_code = []
for i in range(math.ceil(k/5)):
    recovery_code.append(recovery_code_raw[(i*5):((i+1)*5)])


print('-'.join(recovery_code))
```

```python
userID = uuid.uuid4().bytes
mek = uuid.uuid4().bytes
usernameUSER = "Azeroth2231"
passwordUSER = "Primaveraadsd"
saltUser = base64.urlsafe_b64encode(uuid.uuid4().bytes)
saltMekKey = base64.urlsafe_b64encode(uuid.uuid4().bytes)
```

```python
hashedUserKey = PasswordHasher().hash(password=passwordUSER, salt=saltUser)
hashedMekKey = hash_password_raw(
    password= passwordUSER.encode(),
    salt=saltMekKey,
    time_cost=3,
    memory_cost= 77777,
    hash_len=32,
)
```

```python
data = mek
aad = userID
key = hashedMekKey
chacha = ChaCha20Poly1305(key)
nonce = os.urandom(12)

ct = chacha.encrypt(nonce, data, aad)
chacha.decrypt(nonce, ct, aad)

print("Encrypted : ",ct, "\ndecrypted : " , chacha.decrypt(nonce, ct, aad))
```

```python
connection = sqlite3.connect("test_DB")
current = connection.cursor()
current.execute("create table user(username, pw_verifier, encrypted_mek_key)")
current.execute("insert into user values (?, ?, ?)", (usernameUSER, hashedUserKey, ct))
connection.commit()
connection.close()
```
